```python
import torch
from transformers import pipeline, set_seed

# Check whether a GPU is available; falls back to CPU automatically
device = 0 if torch.cuda.is_available() else -1
print("Using GPU" if device == 0 else "Using CPU")
```

```python
generator = pipeline("text-generation", model="gpt2", device=device)
set_seed(42)
print("Model loaded successfully.")
```

```python
def generate_text(prompt, max_length=60, num_return_sequences=1):
    """Generate text continuing the given prompt using the loaded model."""
    outputs = generator(
        prompt,
        max_length=max_length,
        num_return_sequences=num_return_sequences,
        do_sample=True,
        top_k=50,
        top_p=0.95,
        temperature=0.9,
        truncation=True,
    )
    return [o["generated_text"] for o in outputs]
```

```python
user_prompt = input("Enter a sentence or topic: ")

result = generate_text(user_prompt)
print("\nGenerated text:\n")
print(result[0])
```

```python
test_prompts = [
    "The future of artificial intelligence is",
    "Once upon a time in a quiet village,",
    "Climate change is one of the biggest challenges because",
]

for i, prompt in enumerate(test_prompts, start=1):
    print(f"--- Input {i}: {prompt!r} ---")
    generated = generate_text(prompt)
    print(generated[0])
    print()
```
